#Data Format Conversion



```python
#import libreries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
# I set a seed so I get the same random data every time I run this cell
random.seed(42)
```

```python
#I create a list of 10 fictiona names
names = ["Aarav", "Emma", "Liam", "Priya", "Noah",
         "Olivia", "Ethan", "Sophia", "Ravi", "Mia"]
```

```python
# I create a list of 10 cities
cities = ["Toronto", "Kitchener", "Waterloo", "Ottawa", "Vancouver",
          "Montreal", "Calgary", "Hamilton", "Guelph", "London"]
```

```python
# I create an empty list to hold all my records
records = []

# I loop 500 times to create 100 users automatically
for i in range(500):
    name = random.choice(names)      # I pick one name from the names list
    city = random.choice(cities)     # I pick one city from the cities list
    age = random.randint(18, 65)
    salary = random.randint(20000, 100000)
    records.append({"Name": name, "Age": age, "City": city, "Salary": salary})
```

```python
# I turn my list of records into a DataFrame (a table)
df = pd.DataFrame(records)
```

```python
#load the some rows
df.head()
```

```python
# I save the table as a CSV file without the index column
df.to_csv("users.csv", index=False)
```

In this step, I compared the file sizes of my CSV and Parquet files using a dataset of 500 fictional users. My CSV file was 11,522 bytes (about 11.3 KB) and my Parquet file was 6,860 bytes (about 6.7 KB), so the Parquet file was about 40% smaller. This is because Parquet stores data by column and compresses similar values together, like the repeated names and cities in my dataset.

I also noticed that when I tested with only 100 rows, the Parquet file was bigger than the CSV. This happened because Parquet saves extra information inside the file, like column names and data types, and with very little data this extra information took more space than the compression saved. As the data grew to 500 rows, the compression started to work better and Parquet became smaller.

When I loaded the Parquet file back into a DataFrame, all 500 rows and 4 columns were there with no missing values. The data types also stayed correct, with Age and Salary as int64 and Name and City as object (text). This shows the data stayed accurate after the conversion. Overall, CSV is easier for people to read, but Parquet is smaller, faster, and better for large datasets.

#Exploring Data Models with SQLite and MongoDB

```python
import sqlite3
import pandas as pd
```

```python
## I connect to a database file
conn = sqlite3.connect("users.db")
cursor = conn.cursor()
```

```python
#create table 
cursor.execute("""
CREATE TABLE IF NOT EXISTS users(
    ID INTGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Age INTEGER,
    City TEXT 
)
""")
```

```python
# I clear the table first so re-running this cell doesn't cause duplicate IDs
cursor.execute("DELETE FROM users")
```

```python
# I prepare 5 user records
users = [
    (1, "Aarav", 28, "Kitchener"),
    (2, "Emma", 34, "Toronto"),
    (3, "Liam", 22, "Waterloo"),
    (4, "Priya", 41, "Ottawa"),
    (5, "Noah", 19, "Guelph")
]
```

```python
# I insert all 5 records at once, using ? placeholders to keep the insert safe
cursor.executemany("INSERT INTO users (ID, Name, Age, City) VALUES (?, ?, ?, ?)", users)
conn.commit() 
```

```python
# I retrieve all records with an SQL query
print("All users:")
print(pd.read_sql_query("SELECT * FROM users", conn))
```

```python
# I run a second query to show filtering: users older than 25
print("\nUsers older than 25:")
print(pd.read_sql_query("SELECT Name, Age, City FROM users WHERE Age > 25", conn))
```

Interpretation:

In this step, I created a relational database in SQLite using Python's built-in sqlite3 library. I made a table called users with four columns: ID, Name, Age, and City. The table has a fixed structure, which means every record must follow the same format, and ID is the primary key so each user has a unique number. I inserted 5 user records using placeholders, which is a safe way to add data, and saved the changes with commit. Then I retrieved the data with SQL queries. The first query shows all 5 users, and the second query uses WHERE to show only users older than 25. This shows how a relational database stores data in organized rows and columns and uses SQL to find exactly what I need.

```python
from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017/", serverSelectionTimeoutMS=3000)
print(client.server_info()["version"])
```

```python
#I create a database and a collection
db = client["user_db"]
collection = db["users"]
```

```python
# I clear the collection first so re-running this cell doesn't create duplicates
collection.delete_many({})
```

```python
# I create 5 user documents.
users = [
    {"ID": 1, "Name": "Aarav", "Age": 28, "City": "Kitchener"},
    {"ID": 2, "Name": "Emma",  "Age": 34, "City": "Toronto", "Email": "emma@example.com"},
    {"ID": 3, "Name": "Liam",  "Age": 22, "City": "Waterloo", "Hobbies": ["soccer", "gaming"]},
    {"ID": 4, "Name": "Priya", "Age": 41, "City": "Ottawa"},
    {"ID": 5, "Name": "Noah",  "Age": 19} 
]
```

```python
# I insert all 5 documents at once
collection.insert_many(users)
```

```python
#retrive all documents
print("ALl users:")
all_users = list(collection.find({}, {"_id":0}))
print(pd.DataFrame(all_users))
```

```python
#filtered query - users older then 25
print("\nUsers older than 25:")
older = list(collection.find({"Age": {"$gt": 25}}, {"_id":0, "Name":1, "Age": 1, "City":1}))
print(pd.DataFrame(older))
```

I compared two data models, relational (SQLite) and document (MongoDB), by working in Jupyter Notebook on my computer. I used the sqlite3 library, which is built into Python, to create a database file called users.db with a users table. For MongoDB, I installed the pymongo library to connect Python to MongoDB Server 8.3.4, and created a database called user_db with a users collection. I also used pandas to display the results as tables, and MongoDB Compass to view my MongoDB data visually.

I inserted the same 5 users into both databases and ran two queries on each: one to show all users and one to show only users older than 25. Both databases returned the same results. SQLite used WHERE Age > 25, while MongoDB used {"Age": {"$gt": 25}}. The queries look different, but they do the same job.

The biggest difference was flexibility. SQLite has a fixed structure, so every row must have the same columns. MongoDB is flexible. I gave one user an extra Email, one a list of hobbies, and one no City, and MongoDB accepted all of them without any changes. SQLite was easier to set up because it needs no server, while MongoDB needed a server install but was easy to use afterward. Overall, SQLite is best for structured data that must stay consistent, like orders or payments, and MongoDB is best for data that changes often or differs from record to record, like user profiles.
